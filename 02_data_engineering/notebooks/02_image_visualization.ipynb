# 🖼️ Computer Vision Letterboxing & Bounding Box Label Conversion

This notebook demonstrates:
1. Aspect-ratio preserving image resize via **letterboxing**.
2. Converting Pascal VOC XML bounding box annotations to normalized YOLO format.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

src_path = Path.cwd().parent / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from image_pipeline import letterbox_image, voc_box_to_yolo

# Create synthetic test image (e.g., 200x400 landscape)
test_img = np.zeros((200, 400, 3), dtype=np.uint8)
test_img[:, :] = [70, 130, 180]  # Steel blue background
test_img[50:150, 100:300] = [220, 20, 60]  # Crimson object box

padded_img, scale, (pad_w, pad_h) = letterbox_image(test_img, target_size=(416, 416))

print(f"Original shape: {test_img.shape}")
print(f"Padded shape: {padded_img.shape}")
print(f"Scale applied: {scale:.3f}")
print(f"Padding: pad_w={pad_w}, pad_h={pad_h}")

# Pascal VOC box for the crimson rectangle: [xmin, ymin, xmax, ymax]
voc_box = (100.0, 50.0, 300.0, 150.0)
yolo_box = voc_box_to_yolo(voc_box, image_width=400, image_height=200)
print(f"YOLO normalized (x_center, y_center, w, h): {yolo_box}")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(test_img)
axes[0].set_title("Original Image (200x400)")
axes[1].imshow(padded_img)
axes[1].set_title("Letterboxed Image (416x416)")
plt.tight_layout()
plt.show()
```
