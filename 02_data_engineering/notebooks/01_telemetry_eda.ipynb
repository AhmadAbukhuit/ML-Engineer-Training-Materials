# 📊 Exploratory Data Analysis & Polars Telemetry Pipeline

In this notebook, we practice extracting, inspecting, and aggregating heterogeneous raw IoT telemetry logs (JSON and XML) using the high-performance **Polars** DataFrame library.

```python
import sys
from pathlib import Path
import polars as pl

# Ensure local src is in path
src_path = Path.cwd().parent / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from telemetry_parser import process_telemetry_directory

raw_dir = Path.cwd().parent.parent / "data" / "raw" / "telemetry"
processed_dir = Path.cwd().parent.parent / "data" / "processed"

print(f"Raw data directory exists: {raw_dir.exists()}")
```

### Run the ETL Pipeline
We parse all JSON and XML logs into a unified, cleaned dataset.

```python
process_telemetry_directory(str(raw_dir), str(processed_dir))
output_csv = processed_dir / "unified_telemetry.csv"

if output_csv.exists():
    df = pl.read_csv(output_csv)
    print("Dataset Schema:")
    print(df.schema)
    print("\nFirst 5 Rows:")
    print(df.head())
    print("\nSummary Statistics:")
    print(df.describe())
```
