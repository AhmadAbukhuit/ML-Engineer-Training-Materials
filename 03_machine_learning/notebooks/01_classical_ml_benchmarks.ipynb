# 📈 Classical Machine Learning Benchmarks: Custom vs Scikit-Learn

In this notebook, we benchmark our from-scratch implementations against industry-standard `scikit-learn` algorithms:
1. **Linear Regression** (Closed-form vs Gradient Descent vs Scikit-Learn).
2. **Logistic Regression** (Decision boundaries & probabilities).
3. **Decision Tree Classifier** (Tree visualization & split boundaries).
4. **K-Means Clustering & Silhouette Validation**.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification, make_blobs
from sklearn.linear_model import LogisticRegression as SklearnLogReg

src_path = Path.cwd().parent / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from linear_models import LogisticRegression as CustomLogReg
from decision_tree import DecisionTreeClassifier as CustomTree
from kmeans import KMeans as CustomKMeans, silhouette_score

print("Modules successfully imported!")
```

### 1. Logistic Regression Benchmark

```python
X, y = make_classification(n_samples=300, n_features=2, n_redundant=0, n_clusters_per_class=1, random_state=42)

custom_clf = CustomLogReg(lr=0.1, epochs=1000)
custom_clf.fit(X, y)
custom_acc = np.mean(custom_clf.predict(X) == y)

sk_clf = SklearnLogReg()
sk_clf.fit(X, y)
sk_acc = np.mean(sk_clf.predict(X) == y)

print(f"Custom Logistic Regression Accuracy: {custom_acc:.3f}")
print(f"Scikit-Learn Logistic Regression Accuracy: {sk_acc:.3f}")
```

### 2. K-Means Clustering & Silhouette Analysis

```python
X_blob, _ = make_blobs(n_samples=300, centers=4, n_features=2, random_state=42)
km = CustomKMeans(n_clusters=4, random_state=42)
labels = km.fit_predict(X_blob)
score = silhouette_score(X_blob, labels)

print(f"Custom K-Means Silhouette Score: {score:.3f}")

plt.figure(figsize=(8, 6))
plt.scatter(X_blob[:, 0], X_blob[:, 1], c=labels, cmap='viridis', s=30, alpha=0.7)
plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], c='red', marker='X', s=200, label='Centroids')
plt.title(f"K-Means Clusters (k=4, Silhouette={score:.3f})")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
